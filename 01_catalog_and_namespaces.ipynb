# 01 · Catalog and namespaces
Creates the `airline` namespace in the `local` Iceberg catalog.
Table names in this project look like `local.airline.flights`.
Safe to re-run.

```python
from pyspark.sql import SparkSession

# Catalog config (REST catalog + MinIO) comes from spark-defaults.conf.
spark = SparkSession.builder.appName("01-catalog").getOrCreate()
spark.sparkContext.setLogLevel("WARN")
print("Spark", spark.version, "| UI:", spark.sparkContext.uiWebUrl)
```

```python
spark.sql("CREATE NAMESPACE IF NOT EXISTS local.airline")
spark.sql("SHOW NAMESPACES IN local").show()
spark.sql("SHOW TABLES IN local.airline").show()
```

Catalog settings Spark is actually using (loaded from `spark-defaults.conf`):

```python
for k in ["type", "uri", "io-impl", "s3.endpoint", "s3.path-style-access"]:
    print(f"spark.sql.catalog.local.{k:<22}", spark.conf.get(f"spark.sql.catalog.local.{k}", "<unset>"))
```
